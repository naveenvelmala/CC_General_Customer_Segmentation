## 📌 Problem Statement

A credit card company wants to design **targeted marketing campaigns** for its customers, but it currently treats its entire customer base as one homogeneous group. This is inefficient — a customer who only ever pays in full and uses the card sparingly has completely different needs from a customer who relies heavily on cash advances and revolves a balance every month.

**Goal:** Segment ~9,000 active credit card holders into distinct behavioral groups based on 18 months of usage data (balance, purchase patterns, cash advances, credit limit, payment behavior, tenure, etc.), using **unsupervised learning**, so the business can:

- Design targeted marketing and retention campaigns per segment
- Identify high-value customers vs. at-risk / high-credit-utilization customers
- Personalize credit limit and rewards offers
- Detect anomalous or outlier spending behavior

Since there are no pre-existing labels for "customer type," this is framed as a **clustering problem**, and we compare two algorithms — **K-Means** (partition-based) and **DBSCAN** (density-based) — to find natural groupings in the data.

## 1. Import Libraries

Standard data science stack: `pandas`/`numpy` for data handling, `matplotlib`/`seaborn` for visualization, and `scikit-learn` for scaling, clustering, dimensionality reduction, and evaluation.

````python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

sns.set_style("whitegrid")
````

## 2. Load the Dataset

If you're running this in **Google Colab**, upload `CC_GENERAL.csv` to the Colab session (or mount Google Drive) so the path below resolves correctly.

````python
df = pd.read_csv("/content/CC GENERAL.csv")

df.head()
````

## 3. Initial Data Exploration

Before touching the data, get a feel for its shape, types, summary statistics, missing values, and duplicate rows.

````python
print("Dataset shape:", df.shape)
````

````python
df.info()
````

````python
df.describe().T
````

````python
df.isnull().sum()
````

````python
print("Duplicate rows:", df.duplicated().sum())
````

## 4. Data Cleaning

- Drop exact duplicate rows
- Set aside `CUST_ID` (not a behavioral feature — clustering algorithms should never see it)
- Impute missing values with the **median** (robust to the outliers/skew common in financial data)


````python
df = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df.shape)
````

````python
customer_ids = df["CUST_ID"].copy()

df = df.drop(columns=["CUST_ID"])
````

````python
missing_before = df.isnull().sum()

missing_before[missing_before > 0]
````

````python
for column in df.columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())
````

````python
df.isnull().sum().sum()
````

````python
df_original = df.copy()
````

## 5. Feature Scaling

K-Means and DBSCAN are both **distance-based** algorithms, so features on larger scales (e.g. `CREDIT_LIMIT` in the thousands) would dominate features on smaller scales (e.g. `PURCHASES_FREQUENCY` between 0 and 1) unless everything is standardized first. We use `StandardScaler` to give every feature zero mean and unit variance.


````python
scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)
````

````python
scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns,
    index=df.index
)

scaled_df.head()
````

## 6. Finding the Optimal Number of Clusters (K-Means)

Two standard techniques for choosing K:

- **Elbow Method** — plot inertia (within-cluster sum of squares) vs. K, and look for the "elbow" where adding more clusters stops giving much improvement
- **Silhouette Score** — measures how well-separated clusters are (ranges from -1 to 1; higher is better)


````python
inertia_values = []

k_values = range(2, 11)

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(scaled_data)

    inertia_values.append(kmeans.inertia_)
````

````python
plt.figure(figsize=(9, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.title("Elbow Method for Selecting K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia / Within-Cluster Sum of Squares")
plt.xticks(list(k_values))
plt.show()
````

````python
silhouette_scores = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(scaled_data)

    score = silhouette_score(
        scaled_data,
        labels
    )

    silhouette_scores.append(score)
````

````python
silhouette_table = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia_values,
    "Silhouette Score": silhouette_scores
})

silhouette_table
````

````python
plt.figure(figsize=(9, 5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for Different K Values")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.xticks(list(k_values))
plt.show()
````

````python
best_k = silhouette_table.loc[
    silhouette_table["Silhouette Score"].idxmax(),
    "K"
]

print("Best K based on silhouette score:", best_k)
````

We select **K = 4** as a practical trade-off: it sits at a clear bend in the elbow plot and produces clusters that are large enough to be actionable for marketing segments, rather than over-fragmenting the customer base purely to chase the single highest silhouette score.


````python
selected_k = 4
````

## 7. K-Means Clustering

Fit the final K-Means model with the chosen K and attach cluster labels back to the (unscaled) dataframe for interpretability.

````python
kmeans_model = KMeans(
    n_clusters=selected_k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans_model.fit_predict(scaled_data)
````

````python
df_original["KMeans_Cluster"] = kmeans_labels
````

````python
kmeans_counts = df_original[
    "KMeans_Cluster"
].value_counts().sort_index()

kmeans_counts
````

````python
kmeans_silhouette = silhouette_score(
    scaled_data,
    kmeans_labels
)

print(
    "K-Means silhouette score:",
    round(kmeans_silhouette, 4)
)
````

## 8. Dimensionality Reduction for Visualization (PCA)

The data has 17 dimensions, which can't be plotted directly. We use **PCA** to project it down to 2 principal components purely for visualization — the clustering itself was done on the full scaled feature space, not on these 2 components.


````python
pca = PCA(n_components=2)

pca_data = pca.fit_transform(scaled_data)
````

````python
pca_df = pd.DataFrame({
    "PCA1": pca_data[:, 0],
    "PCA2": pca_data[:, 1],
    "KMeans_Cluster": kmeans_labels
})
````

````python
print(
    "Explained variance ratio:",
    pca.explained_variance_ratio_
)

print(
    "Total variance explained by two components:",
    round(pca.explained_variance_ratio_.sum(), 4)
)
````

````python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=pca_df,
    x="PCA1",
    y="PCA2",
    hue="KMeans_Cluster",
    palette="tab10",
    s=45,
    alpha=0.7
)

plt.title("K-Means Customer Clusters After PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="K-Means Cluster")
plt.show()
````

## 9. Cluster Profiling (K-Means)

Numbers alone don't tell a marketing team anything — we need to describe **who** is in each cluster. We compute the mean of each key feature per cluster, then express it **relative to the overall average** so it's easy to see which behaviors define each segment (e.g. "2.5x the average cash advance usage").


````python
kmeans_profile = df_original.groupby(
    "KMeans_Cluster"
).mean(numeric_only=True)

kmeans_profile.round(2)
````

````python
important_features = [
    "BALANCE",
    "PURCHASES",
    "ONEOFF_PURCHASES",
    "INSTALLMENTS_PURCHASES",
    "CASH_ADVANCE",
    "CREDIT_LIMIT",
    "PAYMENTS",
    "MINIMUM_PAYMENTS",
    "PURCHASES_FREQUENCY",
    "CASH_ADVANCE_FREQUENCY",
    "PRC_FULL_PAYMENT",
    "TENURE"
]

kmeans_summary = (
    df_original
    .groupby("KMeans_Cluster")[important_features]
    .mean()
    .round(2)
)

kmeans_summary
````

````python
kmeans_summary["Customer_Count"] = (
    df_original
    .groupby("KMeans_Cluster")
    .size()
)

kmeans_summary
````

````python
overall_average = df_original[
    important_features
].mean()

relative_kmeans_profile = (
    kmeans_summary[important_features]
    .divide(overall_average)
)

relative_kmeans_profile.round(2)
````

## 10. DBSCAN Clustering

As an alternative to K-Means, we try **DBSCAN**, a density-based algorithm. Unlike K-Means, DBSCAN doesn't require specifying the number of clusters up front, can find arbitrarily-shaped clusters, and explicitly flags **outliers/noise points** (label `-1`) — useful for spotting unusual spending behavior.


````python
dbscan_model = DBSCAN(
    eps=1.8,
    min_samples=10
)

dbscan_labels = dbscan_model.fit_predict(scaled_data)
````

````python
unique_labels, counts = np.unique(
    dbscan_labels,
    return_counts=True
)

dbscan_counts = pd.DataFrame({
    "Cluster": unique_labels,
    "Customer_Count": counts
})

dbscan_counts
````

````python
number_of_clusters = len(
    set(dbscan_labels) - {-1}
)

number_of_noise_points = np.sum(
    dbscan_labels == -1
)

print("Number of DBSCAN clusters:", number_of_clusters)
print("Number of noise points:", number_of_noise_points)
````

````python
non_noise_mask = dbscan_labels != -1

non_noise_labels = dbscan_labels[
    non_noise_mask
]

if len(np.unique(non_noise_labels)) >= 2:

    dbscan_silhouette = silhouette_score(
        scaled_data[non_noise_mask],
        non_noise_labels
    )

    print(
        "DBSCAN silhouette score:",
        round(dbscan_silhouette, 4)
    )

else:
    dbscan_silhouette = np.nan

    print(
        "Silhouette score cannot be calculated because "
        "DBSCAN produced fewer than two non-noise clusters."
    )
````

````python
dbscan_pca_df = pd.DataFrame({
    "PCA1": pca_data[:, 0],
    "PCA2": pca_data[:, 1],
    "DBSCAN_Cluster": dbscan_labels
})
````

````python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=dbscan_pca_df,
    x="PCA1",
    y="PCA2",
    hue="DBSCAN_Cluster",
    palette="tab10",
    s=45,
    alpha=0.7
)

plt.title("DBSCAN Customer Clusters After PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="DBSCAN Cluster")
plt.show()
````

````python
df_original["DBSCAN_Cluster"] = dbscan_labels
````

````python
dbscan_summary = (
    df_original
    .groupby("DBSCAN_Cluster")[important_features]
    .mean()
    .round(2)
)

dbscan_summary["Customer_Count"] = (
    df_original
    .groupby("DBSCAN_Cluster")
    .size()
)

dbscan_summary
````

## 11. DBSCAN Hyperparameter Tuning

DBSCAN is sensitive to `eps` (the neighborhood radius). We sweep a range of `eps` values to see how the number of clusters, noise percentage, and silhouette score trade off against each other, which justifies the `eps=1.8` choice used above.


````python
dbscan_experiments = []

for eps_value in [
    0.8,
    1.0,
    1.2,
    1.5,
    1.8,
    2.0,
    2.5
]:

    model = DBSCAN(
        eps=eps_value,
        min_samples=10
    )

    labels = model.fit_predict(scaled_data)

    clusters = len(set(labels) - {-1})
    noise_points = np.sum(labels == -1)

    non_noise = labels != -1

    if clusters >= 2:
        score = silhouette_score(
            scaled_data[non_noise],
            labels[non_noise]
        )
    else:
        score = np.nan

    dbscan_experiments.append({
        "eps": eps_value,
        "Number of Clusters": clusters,
        "Noise Points": noise_points,
        "Noise Percentage": (
            noise_points / len(labels)
        ) * 100,
        "Silhouette Score": score
    })
````

````python
dbscan_parameter_results = pd.DataFrame(
    dbscan_experiments
)

dbscan_parameter_results.round(4)
````

## 12. Model Comparison: K-Means vs. DBSCAN

Side-by-side comparison on cluster count, noise points, and silhouette score.

````python
comparison_results = pd.DataFrame({
    "Algorithm": [
        "K-Means",
        "DBSCAN"
    ],

    "Number of Clusters": [
        len(np.unique(kmeans_labels)),
        len(set(dbscan_labels) - {-1})
    ],

    "Noise Points": [
        0,
        np.sum(dbscan_labels == -1)
    ],

    "Silhouette Score": [
        kmeans_silhouette,
        dbscan_silhouette
    ]
})

comparison_results.round(4)
````

## 13. Business Insights from K-Means Clusters

For each K-Means cluster, we print the top 3 features that are **most above** the overall average and the top 3 that are **most below** average — this is what turns raw cluster numbers into a segment description a marketing or credit-risk team can actually act on.


````python
relative_profile = (
    df_original
    .groupby("KMeans_Cluster")[important_features]
    .mean()
    .divide(
        df_original[important_features].mean()
    )
)

for cluster_number in relative_profile.index:

    cluster_values = relative_profile.loc[
        cluster_number
    ]

    highest_features = (
        cluster_values
        .sort_values(ascending=False)
        .head(3)
    )

    lowest_features = (
        cluster_values
        .sort_values()
        .head(3)
    )

    print(f"\nK-Means Cluster {cluster_number}")

    print("\nHighest relative characteristics:")
    print(highest_features.round(2))

    print("\nLowest relative characteristics:")
    print(lowest_features.round(2))
````

## 14. Conclusion & Next Steps

- **K-Means (K=4)** produced compact, well-separated, evenly-sized segments and is the more directly actionable model for marketing — every customer gets assigned to exactly one segment.
- **DBSCAN** is valuable as a complementary check: it naturally isolates outlier customers (noise points) that don't fit any clear behavioral pattern, which K-Means is forced to assign to the "nearest" cluster regardless of fit.
- The relative-feature analysis above lets each K-Means cluster be given a human-readable name (e.g. *"Revolvers,"* *"Transactors who pay in full,"* *"Cash-advance heavy,"* *"Low-engagement/dormant"*) that a marketing or credit-risk team can build campaigns around.

**Possible extensions:**
- Try Gaussian Mixture Models or Hierarchical Clustering for comparison
- Engineer additional ratio features (e.g. `PAYMENTS / CREDIT_LIMIT`) before clustering
- Validate segments against real campaign response or churn data, if available
- Build a lightweight classifier to assign *new* customers to a segment in real time
